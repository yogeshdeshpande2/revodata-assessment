# PySpark Data Validation Example

This notebook demonstrates basic data validation checks using PySpark, including filtering and counting 'Yes' values in a DataFrame.

```python
# Import Required Libraries
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
# Initialize Spark Session
spark = SparkSession.builder.appName("DataValidationExample").getOrCreate()
```

```python
# Create DataFrame with Yes/No Values
data = [(1, "Yes"), (2, "No"), (3, "Yes"), (4, "No"), (5, "Yes")]
columns = ["id", "response"]
df = spark.createDataFrame(data, columns)
df.show()
```

```python
# Filter Rows with 'Yes'
df_yes = df.filter(col("response") == "Yes")
df_yes.show()
```

```python
# Count 'Yes' Occurrences
count_yes = df_yes.count()
print(f"Number of 'Yes' responses: {count_yes}")
```
